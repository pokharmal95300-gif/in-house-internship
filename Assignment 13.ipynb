```python
# Q1) Replace Nan with 0 and Interchange 3 rows and 3 columns of 2D array [[6, -8, 73, -110], [np.nan, -8, 0, 94]]
import numpy as np

arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])

arr = np.nan_to_num(arr, nan=0)
print(arr)
arr = arr.T
print(arr)
```

```python
# Q2) Move axes of 3D array to new positions
import numpy as np

arr = np.arange(24).reshape(2, 3, 4)
new_arr = np.moveaxis(arr, 0, 2)
print(new_arr)
```

```python
# Q3) Replace NaN values with average of columns
import numpy as np

arr = np.array([[10, np.nan, 30],
                [40, 50, np.nan],
                [70, 80, 90]])

col_mean = np.nanmean(arr, axis=0)
inds = np.where(np.isnan(arr))
arr[inds] = np.take(col_mean, inds[1])
print(arr)
```

```python
# Q4) Replace negative value with zero in numpy array using replace
import numpy as np

arr = np.array([10, -5, 20, -8, 30])
arr[arr < 0] = 0
print(arr)
```

```python
# Q5) Study the following program import numpy as np # create a numpy 1d-arrays arr1 = np.array([3, 4]) arr2 = np.array([1, 0]) # find average of NumPy arrays avg = (arr1 + arr2) / 2 print("Average of NumPy arrays:\n", avg) -> Calculate average mean median mode values of two NumPy 2d-arrays
import numpy as np
from scipy import stats

arr1 = np.array([[2, 4],
                 [6, 8]])

arr2 = np.array([[1, 3],
                 [5, 7]])

avg = (arr1 + arr2) / 2
print("Average Array:")
print(avg)
print("Median =", np.median(avg))
print("Mode =", stats.mode(avg, axis=None))
```

```python
# Q6) Solve the following equation using linalg() and inverse Matrix method x - 2y + 3z = 9 -x + 3y - z = -6 2x - 5y + 5z = 17 
import numpy as np

A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([9, -6, 17])
res = np.linalg.solve(A, B)

print("x =", res[0])
print("y =", res[1])
print("z =", res[2])
```

```python
# Q7) Using "Customizing Matplotlib Visualizations" discussed in Numpy session4 notes plot graph to compare your at least 2 semester result
import numpy as np
import matplotlib.pyplot as plt

subjects = ['Maths', 'Python', 'DSA', 'DBMS', 'OS']

sem1 = [78, 82, 75, 80, 77]
sem2 = [85, 88, 83, 86, 84]

x = np.arange(len(subjects))

plt.figure(figsize=(8,5))

plt.plot(x, sem1,
         marker='o',
         linestyle='--',
         label='Semester 1')

plt.plot(x, sem2,
         marker='s',
         linestyle='-',
         label='Semester 2')

plt.title("Semester Result Comparison")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.xticks(x, subjects)
plt.legend()
plt.grid(True)

plt.show()
```

```python
# 1) Practice print statement: - with multiple object - with seperator - with "ends" 
# 1) Practice Print Statement
#a) Print with Multiple Objects
print("Hello", end=" ")
print("World")

print("Welcome", end="---")
print("Python")
```

```python
# b) Print with Separator (sep)
print("2026", "05", "27", sep="-")

print("Python", "Java", "C++", sep=" | ")
```

```python
# c) Print with End (end)
print("Hello", end=" ")
print("World")

print("Welcome", end="---")
print("Python")
```

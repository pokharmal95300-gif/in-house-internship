```python
# Q1) Study the open weather API show more data in your API calling program
import requests

def weather_data(city):
    api_key = "30191583f74e28e5a5e1ea4c4f4fef93"
    url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric"
    
    try:
        response = requests.get(url)
        data = response.json()

        if data["cod"] != 200:
            print("City not found!")
            return

        print("\n--- Weather Report ---")
        print("City:", data["name"])
        print("Temperature:", data["main"]["temp"], "°C")
        print("Feels Like:", data["main"]["feels_like"], "°C")
        print("Humidity:", data["main"]["humidity"], "%")
        print("Pressure:", data["main"]["pressure"], "hPa")
        print("Wind Speed:", data["wind"]["speed"], "m/s")
        print("Weather:", data["weather"][0]["description"])

    except requests.exceptions.RequestException as e:
        print("Error:", e)

city = input("Enter city name: ")
weather_data(city)
```

```python
# Q2) Try building today's game yourself
import random

choices = ["rock", "paper", "scissors"]

while True:
    try:
        c = int(input('''\nGame Start......
1 YES
2 NO | Exit
Enter choice: '''))

        if c == 1:
            uscore = 0
            cscore = 0

            for round in range(1, 6):
                print(f"\n--- Round {round} ---")

                try:
                    userinput = int(input('''
1 ROCK
2 PAPER
3 SCISSORS
Enter your choice: '''))

                    if userinput == 1:
                        uch = "rock"
                    elif userinput == 2:
                        uch = "paper"
                    elif userinput == 3:
                        uch = "scissors"
                    else:
                        print("Invalid choice! Try again.")
                        continue

                except ValueError:
                    print("Please enter number only!")
                    continue

                cch = random.choice(choices)

                print("Your Choice:", uch)
                print("Computer Choice:", cch)

                if uch == cch:
                    print("Draw!")
                    uscore += 1
                    cscore += 1
                elif ((uch == "rock" and cch == "scissors") or
                      (uch == "paper" and cch == "rock") or
                      (uch == "scissors" and cch == "paper")):
                    print("You Win this round!")
                    uscore += 2
                else:
                    print("Computer Wins this round!")
                    cscore += 2

                print("Score -> You:", uscore, "| Computer:", cscore)

            print("\n--- Final Result ---")
            if uscore == cscore:
                print("Game Drawn!")
            elif uscore > cscore:
                print("🎉 You Won the Game!")
            else:
                print("💻 Computer Won the Game!")

        elif c == 2:
            print("Game Exit. Thank you!")
            break

        else:
            print("Invalid menu choice!")

    except ValueError:
        print("Enter only numbers (1 or 2)!")
```

```python
# (B) Random User API
import requests

url = "https://randomuser.me/api/"
response = requests.get(url)
data = response.json()

user = data["results"][0]

print("Name:", user["name"]["first"], user["name"]["last"])
print("Email:", user["email"])
print("Country:", user["location"]["country"])
```

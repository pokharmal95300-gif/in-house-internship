```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
    .appName("Partition Management") \
    .getOrCreate()

print("Spark Session Created")
```

```python
# Generate 5 Million Records
df = spark.range(5000000)

print("Total Records:")
print(df.count())
```

```python
# Initial Partitions
print("Initial Partitions:")
print(df.rdd.getNumPartitions())
```

```python
# Repartition to 12
df_repartition = df.repartition(12)
print("Partitions After Repartition:")
print(df_repartition.rdd.getNumPartitions())
```

```python
# Coalesce to 3
df_coalesce = df_repartition.coalesce(3)

print("Partitions After Coalesce:")
print(df_coalesce.rdd.getNumPartitions())
```

```python
# Display Sample Records
df_coalesce.show(10)
```

```python
spark.stop()
```

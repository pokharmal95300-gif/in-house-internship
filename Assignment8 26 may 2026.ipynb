```python
# 1>Perform an inner merge on this common column and display the resulting DataFrame.
import pandas as pd

df1 =pd.DataFrame({
    'ID': [1, 2 , 3],
    'Name': ['A', 'B', 'C']
        })
df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 95]
})

result = pd.merge(df1, df2, on='ID', how='inner')
print(result)
```

```python
# 2>Perform a left join of df1 and df2 on the 'ID' column. Explain how missing values are handled in the resulting DataFrame. Right Join and Index-Based Join.
# a. Left Join
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['A', 'B', 'C', 'D']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 95]})

left_join = pd.merge(df1, df2, on='ID', how='left')
print(left_join)
```

```python
# b. Right Join
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['A', 'B', 'C', 'D']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 95]})

right_join = pd.merge(df1, df2, on='ID', how='right')
print(right_join)
```

```python
# c. Index-Based Join
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['A', 'B', 'C', 'D']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 95]})

df1_index = df1.set_index('ID')
df2_index = df2.set_index('ID')

index_join = df1_index.join(df2_index, how='left')
print(index_join)
```

```python
# 3 >Perform a right join using pd.merge() on a common column, then perform a join using df.join() based on the index. Compare the results. Merging with Multiple Keys.
# a. Right Join using pd.merge()
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Marks': [90, 95, 85]})

right_join = pd.merge(df1, df2, on='ID', how='right')
print(right_join)
```

```python
# b. Index-Based Join using df.join()
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['A', 'B', 'C']
})

df2 = pd.DataFrame({
    'ID': [2, 3, 4],
    'Marks': [90, 95, 85]})

df1_index = df1.set_index('ID')
df2_index = df2.set_index('ID')

index_join = df1_index.join(df2_index, how='right')
print(index_join)
```

```python
# c. Merging with Multiple Keys
df1 = pd.DataFrame({
    'ID': [1, 1, 2],
    'Subject': ['Math', 'Science', 'Math'],
    'Marks1': [80, 85, 90]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 2],
    'Subject': ['Math', 'Math', 'Science'],
    'Marks2': [88, 92, 95]
})

multi_merge = pd.merge(df1, df2, on=['ID', 'Subject'], how='inner')
print(multi_merge)
```

```python
# Q3) Create three DataFrames. Vertically concatenate two of them using pd.concat(), then merge the resulting DataFrame with the third DataFrame on a common key. T Understand join() vs. merge().
# a. Vertical Concatenation
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['A', 'B']})

df2 = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['C', 'D']})

df3 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [80, 85, 90, 95]})

concat_df = pd.concat([df1, df2], axis=0)
print(concat_df)
```

```python
#b. Merge with Third DataFrame
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['A', 'B']})

df2 = pd.DataFrame({
    'ID': [3, 4],
    'Name': ['C', 'D']})

df3 = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Marks': [80, 85, 90, 95]})

result = pd.merge(concat_df, df3, on='ID', how='inner')
print(result)
```

```python
# Also Explain the primary differences between df.join() and pd.merge(),
Answer: Primary Differences between df.join() and pd.merge()
df.join() and pd.merge() are both used to combine DataFrames, but they work in different ways.

Basis of joining:
pd.merge() performs joining based on one or more columns (e.g., a common column like ID).
df.join() performs joining based on the index of the DataFrames.

Flexibility:
merge() is more flexible because it allows joining on multiple columns and supports different join conditions.
join() is simpler and mainly used for index-based operations.

Syntax:
merge() requires specifying the column using on='column_name'.
join() usually works directly on the index or after setting a column as index.
    
Use case:
Use merge() when you want to combine data using common columns.
Use join() when the index already represents meaningful keys.
```

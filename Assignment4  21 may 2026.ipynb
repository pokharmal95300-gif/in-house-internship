```python
# 1. Write a Python function to find the maximum of three numbers.
def max_of_three(a, b, c):
    return max(a, b, c)

print(max_of_three(10, 25, 15))
```

```python
# 2. Write a Python function that takes a list and returns a new list with distinct elements.
def distinct_list(lst):
    return list(set(lst))

print(distinct_list([1, 2, 2, 3, 4, 4]))
```

```python
# 3. Write a Python function to multiply all the numbers in a list.
def multiply_list(lst):
    result = 1
    for i in lst:
        result *= i
    return result

print(multiply_list([1, 2, 3, 4]))
```

```python
# 4. Write a Python function to calculate the factorial of a number.
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact
print(factorial(5))    
```

```python
# 5. Write a Python program to reverse a string.
def reverse_string(s):
    return s[::-1]

print(reverse_string("hello"))
```

```python
# 6. Write a Python function to check whether a number falls within a given range.
def check_range(num, start, end):
    return start <= num <= end

print(check_range(15, 1, 25))
```

```python
# 7. Write a Python function to print even numbers from a given list.
def even_numbers(lst):
    for i in lst:
        if i % 2 == 0:
            print(i)

even_numbers([1, 2, 3, 4, 5, 6])
```

```python
# 8. Write a Python function to check whether a number is prime or not.
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

print(is_prime(7))
```

```python
# 9. Write a Python function to count upper and lower case letters.
def count_case(s):
    upper = 4
    lower = 5

    for ch in s:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1

    print("Uppercase:", upper)
    print("Lowercase:", lower)

count_case("Hello World")
```

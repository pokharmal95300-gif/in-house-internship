```python
# Q1) Create numpy array and perform following operation
# 1. Convert 1D array to 2D
import numpy as np
arr1 = np.array([1, 2, 3, 4, 5, 6])
arr2d = arr1.reshape(2, 3)
print(arr2d)
```

```python
# 2. Print Array Attributes(Like shape, dimenssion, data type, itemsize)
print("Shape:", arr2d.shape)
print("Dimension:", arr2d.ndim)
print("Data Type:", arr2d.dtype)
print("Item Size:", arr2d.itemsize)
```

```python
# 3. Create a 3×3 NumPy array of all 9
arr = np.full((3, 3), 9)
print(arr)
```

```python
# 4. Create a 1D array of 10 evenly spaced values between 25 and 125
arr = np.linspace(25, 125, 10)
print(arr)
```

```python
# 5. Convert a Python list into a NumPy array
lst = [10, 20, 30, 40]
arr = np.array(lst)
print(arr)
```

```python
# 6. Reverse a 1D NumPy array
arr = np.array([1, 2, 3, 4, 5])
print(arr[::-1])
```

```python
# 7. Create a 4×4×3 array and extract value at second set, first row and last column
arr = np.arange(48).reshape(4, 4, 3)
print(arr)
print(arr[1, 0, 2])
```

```python
# 8. Create a 4×4 and Extract Odd Rows and Even Columns
arr = np.arange(1, 17).reshape(4, 4)
print(arr)
print(arr[::2, 1::2])
```

```python
# 9. Slice first two rows and first two columns of second set from 4×4×3 array
arr = np.arange(48).reshape(4, 4, 3)
print(arr[1, :2, :2])
```

```python
# 10. Replace all odd numbers with -1 using for loop
arr = np.array([[23, 56, 78, 93],
                [71, 82, 13, 24]])

for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        if arr[i][j] % 2 != 0:
            arr[i][j] = -1
print(arr)            
```

```python
# 11. Get indices of non-zero elements
arr = np.array([1, 0, 2, 0, 3, 0, 4])
print(np.nonzero(arr))
```

```python
# 12. Arithmetic Operations on Two Arrays
arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2, 3])
print(arr1 + arr2)
print(arr1 * arr2)
```

```python
# 13. Dot Product
arr1 = np.array([15, 20, 25])
arr2 = np.array([10, 40, 37])

dot_product = np.dot(arr1, arr2)
print(dot_product)
```

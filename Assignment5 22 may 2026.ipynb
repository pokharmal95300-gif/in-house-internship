```python
#Q2) Practice DATABASE

#(1) Create Database
import sqlite3
conn = sqlite3.connect("college.db")
```

```python
#(2) Create 2–3 Tables
#1 faculty table
conn.execute("""
CREATE TABLE faculty(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name VARCHAR(50),
    age INTEGER
)
""")
```

```python
#2 course Table 
conn.execute("""
CREATE TABLE IF NOT EXISTS course(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    branch VARCHAR(10),
    course_name VARCHAR(10)
)
""")
```

```python
#3) Insert some records
#insert record faculty table
conn.execute("INSERT INTO faculty(name, age) VALUES('Ram', 20)")
conn.execute("INSERT INTO faculty(name, age) VALUES('Shyam', 22)")

conn.commit()
```

```python
#insert record course table
conn.execute("INSERT INTO course(branch, course_name) VALUES('CSE', 'Python')")
conn.execute("INSERT INTO course(branch, course_name) VALUES('IT', 'Java')")

conn.commit()
```

```python
#5) Update some data
conn.execute("UPDATE faculty SET age = 25 WHERE name = 'Ram'")
conn.commit()
```

```python
#5) Update some data
conn.execute("UPDATE course SET course_name = 'Advanced Python' WHERE branch = 'CSE'")
conn.commit()
```

```python
#6) Delete some data
conn.execute("DELETE FROM faculty WHERE name = 'Shyam'")
conn.commit()
```

```python
#6) Delete some data
conn.execute("DELETE FROM course WHERE course_name = 'rahul'")
conn.commit()
```

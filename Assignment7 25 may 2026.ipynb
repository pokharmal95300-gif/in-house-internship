```python
# Q3)Data iteration
# Different ways to iterate over rows in Pandas Dataframe 
# 1. Using iterrows()
import pandas as pd

data = {
    "sub": ["sub1", "sub2", "sub3"],
    "qmark": [90, 80, 85],
    "duration": [45, 60, 50],
}

df = pd.DataFrame(data)

for index, row in df.iterrows():
    print(index, row["sub"], row["qmark"], row["duration"])
```

```python
# 2. Using itertuples()
import pandas as pd

data = {
    "sub": ["sub1", "sub2", "sub3"],
    "qmark": [90, 80, 85],
    "duration": [45, 60, 50],
}

df = pd.DataFrame(data)

for row in df.itertuples():
    print(row.Index, row.sub, row.qmark, row.duration)
```

```python
# 3. Using iloc
import pandas as pd

data = {
    "sub": ["sub1", "sub2", "sub3"],
    "qmark": [90, 80, 85],
    "duration": [45, 60, 50],
}

df = pd.DataFrame(data)

for i in range(len(df)):
    print(df.iloc[i, 0], df.iloc[i, 1], df.iloc[i, 2])
```

```python
# 4. Using loc
import pandas as pd

data = {
    "sub": ["sub1", "sub2", "sub3"],
    "qmark": [90, 80, 85],
    "duration": [45, 60, 50],
}

df = pd.DataFrame(data)

for i in df.index:
    print(df.loc[i, "sub"], df.loc[i, "qmark"], df.loc[i, "duration"])
```

```python
# 5. Using apply()
import pandas as pd

data = {
    "sub": ["sub1", "sub2", "sub3"],
    "qmark": [90, 80, 85],
    "duration": [45, 60, 50],
}

df = pd.DataFrame(data)

def show(row):
    print(row["sub"], row["qmark"], row["duration"])

df.apply(show, axis=1)
```

```python
# Selecting rows in pandas DataFrame based on conditions 
import pandas as pd

df = pd.DataFrame(
    [['a', 'b', 1, 2],
     ['c', 'd', 34, 35],
     ['e', 'f', 5, 6],
     ['g', 'h', 7, 8]],
    columns=['col1', 'col2', 'col3', 'col4']
)

df.index = ['r1', 'r2', 'r3', 'r4']
res = df[df['col3'] != 0]
print(res)
```

```python
# Limited rows selection with given column
import pandas as pd

df = pd.DataFrame(
    [['a', 'b', 1, 2], ['c', 'd', 34, 35],['e', 'f', 5, 6],['g', 'h', 7, 8]],columns=['col1', 'col2', 'col3', 'col4'])

df.index = ['r1', 'r2', 'r3', 'r4']

res = df.loc['r2':'r4', 'col3']
print(res)
```

```python
# Drop rows from the dataframe based on certain condition applied on a column 
import pandas as pd

df = pd.DataFrame(
    [['a', 'b', 1, 2], ['c', 'd', 34, 35],['e', 'f', 5, 6],['g', 'h', 7, 8]],columns=['col1', 'col2', 'col3', 'col4'])

print(df)
res = df[df['col3'] <= 5]
print(res)
```

```python
# Insert row at given position in Pandas Dataframe 
import pandas as pd

df = pd.DataFrame(
    [['a', 'b', 1, 2], ['c', 'd', 34, 35],['e', 'f', 5, 6],['g', 'h', 7, 8]],columns=['col1', 'col2', 'col3', 'col4'])

print(df)
new_row = ['i', 'j', 9, 10]
pos = 2
new_df = pd.DataFrame([new_row], columns=df.columns)
df = pd.concat([df.iloc[:pos], new_df, df.iloc[pos:]]).reset_index(drop=True)
print(df)
```

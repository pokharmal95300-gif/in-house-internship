```python
# 1. Repeat a tuple three times using * operator
t = (1, 2, 3)
print("1.", t * 3)
```

```python
# 2. Join three tuples using + operator
t1 = (1, 2)
t2 = (3, 4)
t3 = (5, 6)
result = t1 + t2 + t3
print("2.", result)
```

```python
# 3. Check whether an element exists in a tuple
numbers = (10, 20, 30, 40, 50)
element = 30
if element in numbers:
    print("3. Element exists in tuple")
else:
    print("3. Element does not exist in tuple")
```

```python
# 4. Find total, highest and lowest without sum(), max(), min()
nums = (5, 10, 15, 20, 25)

total = 0
highest = nums[0]
lowest = nums[0]

for n in nums:
    total += n

    if n > highest:
        highest = n

    if n < lowest:
        lowest = n

print("4. Total =", total)
print("   Highest =", highest)
print("   Lowest =", lowest)
```

```python
# 5. Filter values greater than 10
values = (3, 14, 7, 22, 9, 41, 18, 5)

filtered = ()

for v in values:
    if v > 10:
        filtered += (v,)

print("5.", filtered)
```

```python
# 6. Count elements in a set without len()
s = {"cat", "dog", "bird", "fish"}

count = 0
for item in s:
    count += 1

print("6. Number of elements =", count)
```

```python
# 7. Combine two sets
set1 = {1, 2, 3}
set2 = {3, 4, 5}

combined = set1.union(set2)

print("7.", combined)
```

```python
# 8. Common elements in two sets
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

common = s1.intersection(s2)

print("8.", common)
```

```python
# 9. Elements in either set but not both
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

sym_diff = s1.symmetric_difference(s2)

print("9.", sym_diff)
```
